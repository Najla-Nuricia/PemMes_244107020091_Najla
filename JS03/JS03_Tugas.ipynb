```python
	import pandas as pd
	import numpy as np
	from sklearn.model_selection import train_test_split
	from sklearn.compose import ColumnTransformer
	from sklearn.preprocessing import OneHotEncoder, StandardScaler
	from sklearn.pipeline import Pipeline
	from sklearn.impute import SimpleImputer
	from sklearn.linear_model import LogisticRegression
	from sklearn.feature_selection import chi2, SelectKBest, RFE
	from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv('wbc.csv')
```

```python
df.head()
```

```python
# Hapus kolom yang tidak digunakan
df = df.drop(columns=['id', 'Unnamed: 32'])

# Pisahkan fitur dan target
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

print(X.shape)
print(y.shape)
```

```python
df['diagnosis'] = df['diagnosis'].map({
    'B': 0,
    'M': 1
})
```

```python
numeric_columns = df.select_dtypes(include=["float64"]).columns

preprocess = ColumnTransformer([
    ("num", StandardScaler(), numeric_columns)
])
```

```python
# Seleksi fitur dengan SelectKBest
from sklearn.feature_selection import f_classif

selector_filter = SelectKBest(
    score_func=f_classif,
    k=5
)
```

```python
pipe_filter = Pipeline([
    ("prep", preprocess),
    ("sel", selector_filter),
    ("clf", LogisticRegression(max_iter=1000))
])
```

```python
	# Lakukan pelatihan dan uji model
	X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)

	pipe_filter.fit(X_train, y_train)
	pred = pipe_filter.predict(X_test)
	print("=== Filter (ANOVA) + LR ===")
	print("Accuracy:", accuracy_score(y_test, pred))
	print(classification_report(y_test, pred))
```

```python
selector = pipe_filter.named_steps["sel"]

selected_features = numeric_columns[selector.get_support()]

print("Fitur terbaik:")
print(selected_features.tolist())
```
