```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv('spam.csv', encoding='latin-1')

df = df[['v1', 'v2']]

df.columns = ['label', 'message']

print(df.head())
print(df.shape)
```

```python
X = df['message']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print('Jumlah data train:', len(X_train))
print('Jumlah data test:', len(X_test))
```

```python
bow = CountVectorizer(stop_words='english')

X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)

print('Jumlah fitur:', len(bow.get_feature_names_out()))
print('Dimensi data train:', X_train_bow.shape)
```

```python
mnb_bow = MultinomialNB()

mnb_bow.fit(X_train_bow, y_train)

y_pred_bow = mnb_bow.predict(X_test_bow)

acc_bow = accuracy_score(y_test, y_pred_bow)

print(f'Akurasi CountVectorizer: {acc_bow:.4f}')
```

```python
print(classification_report(y_test, y_pred_bow))
```

```python
print(confusion_matrix(y_test, y_pred_bow))
```

berdasarkan hasil tingkat keakuratan yaitu 0.9844

dan confusion matrix untuk data test menghasilkan TN = 1437, FP = 11, FN = 15, dan TP = 209. Artinya, 1437 data negatif diprediksi benar, 11 data negatif salah diprediksi positif, 15 data positif salah diprediksi negatif, dan 209 data positif diprediksi benar.

```python
# TF-IDF dengan stop_words
tfidf = TfidfVectorizer(stop_words='english')

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print('Jumlah fitur:', len(tfidf.get_feature_names_out()))
print('Dimensi data train:', X_train_tfidf.shape)
```

```python
# Inisiasi Multinomial Naive Bayes
mnb_tfidf = MultinomialNB()

# Training
mnb_tfidf.fit(X_train_tfidf, y_train)

# Prediksi
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)

# Evaluasi
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)

print(f'Akurasi TF-IDF: {acc_tfidf:.4f}')
```

```python
print(classification_report(y_test, y_pred_tfidf))
```

```python
confusion_matrix(y_test, y_pred_tfidf)
```

Untuk TF-IDF, confusion matrix menunjukkan TN = 1448, FP = 0, FN = 53, dan TP = 171. Artinya, 1448 data negatif diprediksi benar, tidak ada data negatif yang salah diprediksi positif, 53 data positif salah diprediksi negatif, dan 171 data positif diprediksi benar.

Berdasarkan hasil pengujian pada dataset spam.csv, model Multinomial Naive Bayes dengan fitur CountVectorizer menghasilkan akurasi sebesar 98%, sedangkan penggunaan TF-IDF menghasilkan akurasi sebesar 96%. Dengan demikian,CountVectorizer menghasilkan performa yang lebih tinggi dibandingkan TF-IDF.
